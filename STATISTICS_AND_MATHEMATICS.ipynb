# ***MEDIAN***

The Median is the middle value after sorting the data.

If there are:

1. Odd observation->  Middle Value
2. Even observation-> Average of two middle values



```python
import pandas as pd
import numpy as np


data = {
    "Employee": [
        "A", "B", "C", "D", "E",
        "F", "G", "H", "I", "J"
    ],
    "Age": [
        22, 25, 28, 30, 35,
        40, 32, 29, 45, 38
    ],
    "Salary": [
        25000, 32000, 40000, 42000, 50000,
        62000, 47000, 43000, 75000, 68000
    ],
    "Experience": [
        1, 2, 4, 5, 8,
        12, 6, 5, 18, 15
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("=" * 50)
print("Employee Dataset")
print("=" * 50)
print(df)

# ==========================================
# Median of Individual Columns
# ==========================================

print("\nMedian of Age:")
print(df["Age"].median())

print("\nMedian of Salary:")
print(df["Salary"].median())

print("\nMedian of Experience:")
print(df["Experience"].median())

# ==========================================
# Median of All Numerical Columns
# ==========================================

print("\n" + "=" * 50)
print("Median of All Numerical Columns")
print("=" * 50)

print(df.median(numeric_only=True))
```
