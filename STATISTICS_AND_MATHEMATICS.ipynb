# ***MEDIAN***

The Median is the middle value after sorting the data.

If there are:

1. Odd observation->  Middle Value
2. Even observation-> Average of two middle values



```python
import pandas as pd
import numpy as np


data = {
    "Employee": [
        "A", "B", "C", "D", "E",
        "F", "G", "H", "I", "J"
    ],
    "Age": [
        22, 25, 28, 30, 35,
        40, 32, 29, 45, 38
    ],
    "Salary": [
        25000, 32000, 40000, 42000, 50000,
        62000, 47000, 43000, 75000, 68000
    ],
    "Experience": [
        1, 2, 4, 5, 8,
        12, 6, 5, 18, 1536
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("=" * 50)
print("Employee Dataset")
print("=" * 50)
print(df)

# ==========================================
# Median of Individual Columns
# ==========================================

print("\nMedian of Age:")
print(df["Age"].median())

print("\nMedian of Salary:")
print(df["Salary"].median())

print("\nMedian of Experience:")
print(df["Experience"].median())

# ==========================================
# Median of All Numerical Columns
# ==========================================

print("\n" + "=" * 50)
print("Median of All Numerical Columns")
print("=" * 50)

print(df.median(numeric_only=True))
```

# ***MEASURE OF DISPERSION***

Measure of dispersion describe the variability or spread of data in a dataset

**RANGE**

The simplest measure of dispersion.

Formula:

Range = Maximum value - Minimum Value

```python
import numpy as np

data = [12, 18, 20, 25, 30]

print("Range =", max(data) - min(data))
```

# ***INTERQUARTILE RANGE (IQR)***

IQR measures the spread of the middle 50% of the data.

It ignores extreme values.

# ***Formula***

IOQ = Q3 - Q1

Where,

1. Q1 = First quartile (25%)
2. Q3 = Third Quartile (75%)

```python
import numpy as np

data = [2, 4, 6, 8, 10, 12, 14, 16]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

# ***QUARTILE DEVIATION (SEMI-IQR)***

Quartile Deviation is half of IQR.

# ***FORMULA***

QD = 2 Q3 - Q1

# ***VARIANCE***

Variance measures the average squared deviation from the mean.

A samll variance means data points are close to the mean, while a large variance means they are widely spread.

**FORMULA**

var(X) = E(X2) - [E(X)]2

Their are two types:

# **POPULATION VARIANCE**

s 2 = ∑( x - x ˉ ) 2 /n - 1

# **SAMPLE VARIANCE**

σ 2 = Σ ( x - μ ) 2 /2

```python
import numpy as np

data = [2, 4, 6, 8]

print("Population Variance =", np.var(data))
```

# ***STANDARD DEVIATION***

Standard deviation is the square root of variance.

It is the most widely used measure of dispersion beacuse it is expressed in the same units as the original data.





```python
import numpy as np

data = [2, 4, 6, 8]

print("Population SD =",np.std(data))
print("Sample SD=",np.std(data,ddof = 1))
```
