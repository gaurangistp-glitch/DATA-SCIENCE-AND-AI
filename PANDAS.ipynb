# ***WHAT IS PANDAS***

Pandas is an open source python library used for data manipulation and data analysis.

# ***USE OF PANDAS***

Read, clean, oraganise, filter, analyse, and manipulate structured data efficiently.

# ***Features of Pandas***

Data Manipulation:

Used to modify, oraganise and transform data easily.

Data Analysis:

Used to analyse data and extract meaningful insights.

Fast Processing:

Hnadles large datasets efficiently with optimized performance

Missing value Handling:

Allows easy detection and handling of missing values.

File Handling:

Can read and write CSV, Excel, JSON, SQL nad many other file formats.

Data Filtering:

Used to select specific rows and coulumns based on conditions.

Data Cleaning:

Stastical Analysis:

# ***CREATING SERIES***

```python
import pandas as pd

marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
```

# ***STRING SERIES***

```python
import pandas as pd

names= pd.Series(["Gaurangi", "Khushi", "Sejal", "Anupam"])

print(names)
```

# ***CUSTOM INDEX***

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 78],
    index=["Math", "Science", "English"]
)

print(marks)
```

# ***SERIES FROM DICTIONARY***

```python
import pandas as pd

student = pd.Series({
    "Name": "Gaurangi",
    "Age": 19,
    "City": "Sitapur"
})

print(student)
```

# ***ACCESS VALUES IN A SERIES***

```python
import pandas as pd

marks = pd.Series([85, 90, 78, 92, 88])

print(marks[0])
print(marks[2])
```
