# ***SELECTING DATA***

```python
import pandas as pd

df = pd.read_csv("/content/Employee.csv")

print(df["Gender"]) # Selecting a Single Column
df[["Education", "City"]] # selecting multiple columns

#Selecting Rows using Index

print(df.iloc[0]) # first row
print(df.iloc[0:5]) # multiple rows
print(df.iloc[-5:]) # last five rows

## Selecting Rows and Columns using loc[]

#loc[] uses row labels (index names) and column names.
# syntax -> df.loc[row_label, column_name]
print(df.loc[0]) # first row
print(df.loc[0:5]) # multiple rows
print(df.loc[-5:]) # last five rows

#Selecting Rows and Columns using iloc[]

#iloc[] uses integer positions.
#syntax -> df.iloc[row_index, column_index]

print(df.iloc[0]) #first row
print(df.iloc[0:5]) #multiple rows
print(df.iloc[-5:]) #last five rows

# Boolean Indexing

#Boolean Indexing returns rows where the condition is True.

# Age greater than 30
df[df["Age"] > 30]

# Age less than 30
df[df["Age"] < 30]
# city is banglore
df[df["City"] == "Bangalore"]

# education is master
df[df["Education"] == "Masters"]

# Filtering Data

df[df["City"] == "Bangalore"] # single condition
df[(df["City"] == "Pune") & # multiple conditions(AND) (
   (df["Age"] > 30)]

df[(df["City"] == "Pune") | # multiple conditions(OR)
   (df["Age"] > 30)]

df[~(df["City"] == "New Delhi")] # NOT condition (changed from Department == Sales)

df[df["City"].isin(["Bangalore", "Pune"])] # using isin()

df[df["Age"].between(25, 35)] # using(between)

df[df["Education"].str.contains("Bachelor", case=False)] # using str.contains()
# using query
df.query("Age > 30")
df.query("City == 'Pune'")
df.query("Age > 30 and ExperienceInCurrentDomain > 3")

df.nlargest(5, "Age") # using nlargest()
df.nsmallest(5, "Age") # using nsmallest()
```

# ***MISSING VALUES IN PANDAS***

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Riya", "Karan", "Priya", np.nan],
    "Age": [22, np.nan, 25, 28, 24],
    "City": ["Delhi", "Mumbai", np.nan, "Lucknow", "Delhi"],
    "Salary": [50000, 60000, np.nan, 75000, 55000]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)
print(df.isnull()) #. used to identify missing value
print(df.isnull().sum()) # count missing value
print(df.notnull()) #. used to identify missing value
# fill with a constant
filled = df.fillna("Unknown")
print(filled)
# fill numeric coulmns
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
# fill multiple columns
df.fillna({
    "Name": "Not Available",
    "City": "Unknown",
    "Salary": 0
}, inplace=True)

print(df)

df.dropna() # removes missing values
# df.replace(old_value, new_value) # used to replace speific values (not limited to missing values)
```

# ***SORTING***

```python
import pandas as pd
data = {
    "Emp ID": [105, 101, 104, 102, 103],
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Karan"],
    "Department": ["HR", "IT", "Finance", "IT", "HR"],
    "Salary": [45000, 60000, 55000, 65000, 50000],
    "Age": [28, 24, 35, 30, 27]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
df.sort_values(by="Name") # sort data based on one or more column values, using 'Name' as an example
print(df.sort_values(by="Age", ascending=True)) # sort in ascending order(default)
print(df.sort_values(by="Salary", ascending=False)) # sort in descending order
print(df.sort_values(by=["Department", "Salary"])) # sort by multiple columns
print(df.sort_index()) # sort by index
```

# ***rename()***

Rename column names

```python
import pandas as pd

data = {
    "Emp ID": [101, 102, 103, 104, 105],
    "Name": ["Aman", "Riya", "Karan", "Priya", "Rahul"],
    "Salary": [50000, 60000, 55000, 70000, 65000]

}

df = pd.DataFrame(data)
print("Original DataFrame")
print(df)

df = df.rename(columns={
    "Emp_Id": "ID",
    "Emp_Name":"Name"
})

print("\nAfter Running")
print(df)
#
```

# ***astype()***

Charge the data type of a column.

```python
import pandas as pd

data = {
    "ID": ["101", "102", "103", "104", "105"],
    "Name": ["Aman", "Riya", "Karan", "Priya", "Rahul"]

}

df = pd.DataFrame(data)
print("Before Conversion")
print(df.dtypes)

df["ID"] = df["ID"].astype(int)

print("\nAfter Running")
print(df)
```

# ***replace()***

Replace values in a DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Karan", "Priya", "Rahul"],
    "Age": [28, 24, 35, 30, 27]

}

df = pd.DataFrame(data)

print(df)

df("City") = df["City"].replace("Delhi", "New Delhi")

print("\nAfter Running the Code\n")

print(df)
```

# ***GroupBy***

```python
import pandas as pd

data = {
    "ID" : ["101", "102", "103"],
    "Salary" : ["45000", "60000", "55000"]
}

df = pd.DataFrame(data)

print("Before Conversion:")
print(df.dtypes)

df["ID"] = df["ID"].astype(int)
df["Salary"] = df["Salary"].astype(float)

print("\nAfter Conversion:")
print(df.dtypes)

print(df)
```
