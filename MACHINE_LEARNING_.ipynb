# ***RANDOM FOREST REGRESSION***

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Dataset
data = {
    "Expression": [1, 2, 3, 4, 5, 6, 7, 8],
    "Salary": [30000, 40000, 50000, 60000, 70000, 80000, 90000, 100000]
}

df = pd.DataFrame(data)

x = df[["Expression"]]
y = df["Salary"]

# Model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42)

# Train Model
model.fit(x, y)

# Prediction
salary = model.predict([[5.5]])
print("Predicted Salary", salary)
```

# ***Decision TREE***

A decision tree is a supervised machine learning algorithim used for both:
1. Classification (Predicted categories)
2. Regression (Predicting numerical values)

It works like a flowchart where:
Each internal node represents a decision based on a feature. Each branch represents the outcome of that decision. Each leaf node represents the final prediction.

# ***IMPORTANT TERMINOLOGIES***

1. Root Node

The first node of the tree.

2. Decision Node

A node where data is split.

3. Leaf Node

Final Prediction

4. Branch

Connection between nodes

5. Parent Node

A node that splits into child nodes.

6. Child Node

Resulting node after splitting.

# ***Decision Tree Example***

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Dataset
data = {
    "Age": [25, 35, 45, 20, 30, 50, 23, 40],
    "Income": [30000, 50000, 60000, 20000, 45000, 80000, 25000, 70000],
    "Buy": [0, 1, 1, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)

# Features and Target
X = df[["Age", "Income"]]
y = df["Buy"]

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Create model
model = DecisionTreeClassifier(random_state=42)

# Train model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Visualising the decision
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(10, 6))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["No", "Yes"],
    filled=True,
    rounded=True
)

plt.show()

# Predicting a new customer
new_customer = [[28, 40000]]

prediction = model.predict(new_customer)

if prediction[0] == 1:
    print("Customer will Buy")
else:
    print("Customer will Not Buy")
```
